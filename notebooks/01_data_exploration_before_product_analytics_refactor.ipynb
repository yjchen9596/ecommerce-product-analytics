### Data Type Observations

The initial schema inspection identified several columns whose loaded data types
do not fully represent their analytical meaning:

- `event_time` is stored as a string but represents a timestamp.
- `product_id` is stored as an integer but represents a product identifier.
- `category_id` is stored as an integer but represents a category identifier.
- `user_id` is stored as an integer but represents a user identifier.

The identifier columns should not be treated as quantitative variables because
operations such as mean, standard deviation, and quartiles have no analytical
meaning for IDs.

```python
df["event_time"] = pd.to_datetime(df["event_time"], utc=True)
df["event_time"].dtype
```

```python
id_columns = ["product_id", "category_id", "user_id"]

df[id_columns] = df[id_columns].astype("string")

df[id_columns].dtypes
```

```python
# updated dataset with correct data types

df.describe(include="all")
```
